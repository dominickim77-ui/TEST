```python
# MAMADOMI 음식 메뉴
menu = ["Empanada", "Asado", "Kimchi", "Milanesa"]

# 전체 메뉴 출력
print(menu)
```

```python
print(menu[0])
print(menu[2])
print(menu[-1])
```

```python
menu.append("Choripan")
print(menu)
```

```python
print(menu[0])
print(menu[2])
print(menu[-1])
```

```python
# 두 번째 메뉴 변경
menu[1] = "Pizza"

# 변경된 메뉴 확인
print(menu)
```

```python
# Kimchi 메뉴 삭제
menu.remove("Kimchi")

# 남은 메뉴 확인
print(menu)
```

```python
# 두 번째 메뉴 삭제
deleted_menu = menu.pop(1)

# 삭제한 메뉴 확인
print("삭제한 메뉴:", deleted_menu)

# 남은 메뉴 확인
print("남은 메뉴:", menu)
```

```python
# 첫 번째부터 두 번째 메뉴까지 가져오기
print(menu[0:2])

# 뒤에서 두 개 메뉴 가져오기
print(menu[-2:])
```

```python
# 메뉴를 하나씩 출력하기
for food in menu:
    print(food)
```
